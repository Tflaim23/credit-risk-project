```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

# Root folder is one up from the notebook 
REPO_ROOT = Path.cwd().parent

DATA_PATH = REPO_ROOT / "data" / "processed" / "lendingclub_modeling_v1.parquet"
OUT_DIR = REPO_ROOT / "reports" / "figures" / "modeling_v1"
OUT_DIR.mkdir(parents=True, exist_ok=True)


# Load data
df = pd.read_parquet(DATA_PATH)

print("Rows:", len(df))
print("Cols:", df.shape[1])
print("Bad rate (y=1):", round(df["y"].mean(), 6))

# PLOT 1: Making sure lendingclub grade correlates with bad rate
if "grade" in df.columns:
    grp = df.groupby("grade")["y"].agg(["count", "mean"]).reset_index()
    grp = grp.sort_values("grade")

    plt.figure(figsize=(8, 4))
    plt.plot(grp["grade"], grp["mean"], marker="o")
    plt.title("Bad rate by grade")
    plt.xlabel("Grade")
    plt.ylabel("Bad rate")
    plt.tight_layout()
    plt.savefig(OUT_DIR / "bad_rate_by_grade.png", dpi=150)
    plt.close()

    print("Saved:", OUT_DIR / "bad_rate_by_grade.png")
else:
    print("grade column not found")

# PLOT 2: distribution of loan_amnt
if "loan_amnt" in df.columns:
    plt.figure(figsize=(8, 4))
    plt.hist(df["loan_amnt"].dropna(), bins=50)
    plt.title("Loan amount distribution")
    plt.xlabel("loan_amnt")
    plt.ylabel("Count")
    plt.tight_layout()
    plt.savefig(OUT_DIR / "loan_amnt_hist.png", dpi=150)
    plt.close()

    print("Saved:", OUT_DIR / "loan_amnt_hist.png")
else:
    print("loan_amnt column not found")

# PLOT 3: distribution of debt to income
if "dti" in df.columns:
    dti_non_missing = df["dti"].dropna()

     # Extreme outliers made the plot hard to read and not provide any practical information about the data
     # So I tested caps until I found one that didn't remove too much data but made the plot readable
     # 60 seemed fine any higher didn't make huge differences and hurts the readability of the plot
    dti_cap = 60
    dti_plot = dti_non_missing[dti_non_missing <= dti_cap]

    plt.figure(figsize=(8, 4))
    plt.hist(dti_plot, bins=50)
    plt.title("DTI distribution capped outliers")
    plt.xlabel("dti")
    plt.ylabel("Count")
    plt.tight_layout()
    plt.savefig(OUT_DIR / "dti_hist.png", dpi=150)
    plt.close()

    print("Saved:", OUT_DIR / "dti_hist.png")
    print("DTI missing %:", round(df["dti"].isna().mean() * 100, 4))
    print("DTI > cap count:", int((dti_non_missing > dti_cap).sum()))
else:
    print("dti column not found")
```
